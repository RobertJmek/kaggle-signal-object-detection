# Signal Object Detection - EDA

Exploratory Data Analysis for the Kaggle competition.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os
from pathlib import Path

DATA_DIR = Path('../data/raw')

# Load labels
train_df = pd.read_csv(DATA_DIR / 'train.csv')
test_df = pd.read_csv(DATA_DIR / 'test.csv')

print(f"Train samples: {len(train_df)}")
print(f"Test samples: {len(test_df)}")
print(f"\nLabel distribution:")
print(train_df['label'].value_counts().sort_index())
```

## Visualize Sample Images

```python
fig, axes = plt.subplots(5, 5, figsize=(12, 12))
axes = axes.flatten()

for label in range(1, 6):
    samples = train_df[train_df['label'] == label].sample(5)
    for i, (_, row) in enumerate(samples.iterrows()):
        img_path = DATA_DIR / 'train' / row['id']
        img = Image.open(img_path)
        axes[(label-1)*5 + i].imshow(img)
        axes[(label-1)*5 + i].set_title(f'Label {label}')
        axes[(label-1)*5 + i].axis('off')

plt.tight_layout()
plt.show()
```

## Image Dimensions

```python
# Check image sizes
sample_img = Image.open(DATA_DIR / 'train' / train_df.iloc[0]['id'])
print(f"Image size: {sample_img.size}")
print(f"Image mode: {sample_img.mode}")

# Check a few more
sizes = []
for img_name in train_df['id'].sample(100):
    img = Image.open(DATA_DIR / 'train' / img_name)
    sizes.append(img.size)

unique_sizes = set(sizes)
print(f"\nUnique image sizes (sample of 100): {unique_sizes}")
```
